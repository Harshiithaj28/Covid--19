daily new cases below 

```python
# Convert Date to datetime
covid["Date"] = pd.to_datetime(covid["Date"])

# Calculate total confirmed cases for each day
daily_confirmed = (
    covid.groupby("Date")["Confirmed"]
    .sum()
    .sort_index()
)

# Calculate new cases per day
daily_new_cases = daily_confirmed.diff()

display(daily_new_cases.head(10))
```

```python
plt.figure(figsize=(14, 6))

plt.plot(daily_new_cases)

plt.title("Daily New COVID-19 Cases in India")
plt.xlabel("Date")
plt.ylabel("New Cases")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)

plt.show()
```

7 day 

```python
daily_cases_7day = daily_new_cases.rolling(window=7).mean()

plt.figure(figsize=(14, 6))

plt.plot(daily_cases_7day)

plt.title("7-Day Moving Average of New COVID-19 Cases")
plt.xlabel("Date")
plt.ylabel("New Cases (7-Day Average)")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)

plt.show()
```

```python
peak_date = daily_cases_7day.idxmax()
peak_cases = daily_cases_7day.max()

print("Peak date:", peak_date)
print("Peak 7-day average:", round(peak_cases))
```

Deaths 

```python
daily_deaths = (
    covid.groupby("Date")["Deaths"]
    .sum()
    .sort_index()
)

daily_new_deaths = daily_deaths.diff()

daily_deaths_7day = daily_new_deaths.rolling(7).mean()

plt.figure(figsize=(14, 6))

plt.plot(daily_deaths_7day)

plt.title("7-Day Moving Average of Daily COVID-19 Deaths")
plt.xlabel("Date")
plt.ylabel("Deaths")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)

plt.show()
```

First wave peek

```python
first_wave = daily_cases_7day.loc[
    "2020-06-01":"2020-12-31"
]

first_peak_date = first_wave.idxmax()
first_peak_cases = first_wave.max()

print("First wave peak date:", first_peak_date)
print("First wave peak (7-day average):", round(first_peak_cases))
```

Second wave peek


```python
second_wave = daily_cases_7day.loc[
    "2021-02-01":"2021-07-31"
]

second_peak_date = second_wave.idxmax()
second_peak_cases = second_wave.max()

print("Second wave peak date:", second_peak_date)
print("Second wave peak (7-day average):", round(second_peak_cases))
```

comparing first and the second wave 

```python
wave_comparison = pd.DataFrame({
    "Wave": ["First Wave", "Second Wave"],
    "Peak Date": [first_peak_date, second_peak_date],
    "Peak 7-Day Average": [
        round(first_peak_cases),
        round(second_peak_cases)
    ]
})

display(wave_comparison)
```

```python
wave_ratio = second_peak_cases / first_peak_cases

print(
    f"The second wave peak was approximately "
    f"{wave_ratio:.1f} times the first wave peak."
)
```

Deaths

```python
death_peak_date = daily_deaths_7day.idxmax()
death_peak = daily_deaths_7day.max()

print("Death peak date:", death_peak_date)
print("Peak 7-day average deaths:", round(death_peak))
```

The second COVID-19 wave in India was substantially larger than the first, with its peak 7-day average of approximately 391,280 reported daily cases being 4.2 times the first-wave peak of approximately 93,199. 

The 7-day average of reported deaths peaked at approximately 4,190 on 24 May 2021.
